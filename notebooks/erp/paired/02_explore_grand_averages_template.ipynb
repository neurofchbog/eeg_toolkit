Exploratory visualization of grand averages. Use this notebook to:
- Identify peak latencies and scalp distributions
- Select ROIs and time windows for confirmatory analysis (notebook 03)

**Requires:** Qt5 backend for interactive plots

```python
%load_ext autoreload
%autoreload 2

import matplotlib
matplotlib.use('Qt5Agg')

from eeg_toolkit import (
    load_config,
    load_grand_averages,
    plot_butterfly,
    plot_topomaps,
    plot_channel_overlay,
    plot_channel_overlay_with_ci,
    plot_roi_overlay,
)

# ── Update these paths ──
cfg     = load_config('../../../configs/your_experiment.yaml')
cfg_erp = load_config('../../../configs/your_erp_analysis.yaml')

# Show available conditions per window
for window in cfg_erp.erp.windows:
    gas = load_grand_averages(cfg, window.name)
    print(f"{window.name}: {list(gas.keys())}")
```

```python
# ── Butterfly plots ──
# ── Update window_name and conditions to match your config ──
plot_butterfly(
    cfg,
    window_name='your_window',          # e.g. 'stimulus', 'cue'
    conditions=['condition_a', 'condition_b', 'a_vs_b'],
    times='peaks',
    save=False,
)
```

```python
# ── Topomaps at key timepoints ──
import numpy as np

# ── Update time range to match your epoch window ──
times = np.arange(0.10, 0.60, 0.05).tolist()   # e.g. 100–550 ms in 50 ms steps

plot_topomaps(
    cfg,
    window_name='your_window',
    times=times,
    conditions=['condition_a', 'condition_b', 'a_vs_b'],
    save=False,
)
```

```python
# ── Channel overlay with 95% CI ──
# ── Update channels to match your montage and region of interest ──
channels_of_interest = ['Pz', 'P3', 'P4', 'Oz', 'O1', 'O2']

plot_channel_overlay_with_ci(
    cfg, cfg_erp,
    window_name='your_window',
    channels=channels_of_interest,
    conditions=['condition_a', 'condition_b'],
    save=False,
)
```

```python
# ── ROI overlay ──
# ── Define ROIs based on what you observed above ──
# These will be the ROIs you commit to in notebook 03.
rois = {
    'roi_name_1': ['Ch1', 'Ch2', 'Ch3'],   # e.g. 'posterior': ['Pz', 'P3', 'P4']
    'roi_name_2': ['Ch4', 'Ch5'],
}

plot_roi_overlay(
    cfg,
    window_name='your_window',
    rois=rois,
    conditions=['condition_a', 'condition_b'],
    with_ci=True,
    save=False,
)
```

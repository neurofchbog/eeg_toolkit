# ERP Paired — 03: Confirmatory Analysis

Extracts mean amplitudes, runs paired statistical tests with FDR correction,
and generates publication-ready figures.

**ROIs and time windows should be committed here based on notebook 02 exploration.**
Do not change them after running statistics.

```python
%load_ext autoreload
%autoreload 2

from eeg_toolkit import (
    load_config,
    extract_mean_amplitudes,
    run_paired_tests,
    plot_results,
)

# ── Update these paths ──
cfg     = load_config('../../../configs/your_experiment.yaml')
cfg_erp = load_config('../../../configs/your_erp_analysis.yaml')

print("Setup OK")
```

```python
# ── Commit ROIs and time windows from notebook 02 exploration ──
# Do NOT change these after running statistics.
rois = {
    'roi_name_1': ['Ch1', 'Ch2', 'Ch3'],   # e.g. 'posterior': ['Pz', 'P3', 'P4']
    'roi_name_2': ['Ch4', 'Ch5'],
}

time_windows = {
    'early': [0.10, 0.25],   # in seconds
    'late':  [0.25, 0.50],
}

# ── Extract mean amplitudes ──
df = extract_mean_amplitudes(
    cfg,
    window_name='your_window',
    rois=rois,
    time_windows=time_windows,
    conditions=['condition_a', 'condition_b'],
    save_csv=True,
    verbose=True,
)

df.head()
```

```python
# ── Paired tests with FDR correction ──
results = run_paired_tests(
    df,
    contrasts=('condition_a', 'condition_b'),
    cfg=cfg,
    window_name='your_window',
    save_csv=True,
    verbose=True,
)

results
```

```python
# ── Publication figure ──
# ── Update colors and layout to match your design ──
colors = {
    'condition_a': '#3D5A80',
    'condition_b': '#E07A5F',
}

fig = plot_results(
    cfg,
    window_name='your_window',
    rois=rois,
    conditions=['condition_a', 'condition_b'],
    stats_df=results,
    colors=colors,
    ylim=None,          # e.g. (-3, 4) in µV; None for auto
    xlim=None,          # e.g. (-0.2, 1.0) in s; None for full epoch
    figsize=(10, 3.5),
    ncols=3,
    dpi=300,
    save_path=None,     # set to a Path object to save .png + .tiff
    alpha_fdr=0.05,
    verbose=True,
)
```

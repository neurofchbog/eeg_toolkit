```python
%load_ext autoreload
%autoreload 2

import matplotlib
matplotlib.use('Qt5Agg')

from eeg_toolkit import (
    load_config,
    load_grand_averages,
    plot_butterfly,
    plot_topomaps,
    plot_channel_overlay,
    plot_channel_overlay_with_ci,
    plot_roi_overlay
)

cfg = load_config('../../../configs/eye_eeg_simul.yaml')
cfg_erp = load_config('../../../configs/erp_analysis.yaml')

# Quick overview of what we have
for window in ['trial', 'cue']:
    gas = load_grand_averages(cfg, window)
    print(f"{window}: {list(gas.keys())}")
```

```python
# Butterfly plots: cue window, both conditions + contrast
plot_butterfly(cfg, window_name='trial',
               conditions=['spatial', 'symbolic', 'spatial_vs_symbolic'],
               times='peaks', save=False)
```

```python
import numpy as np

# Sample timepoints from 200 to 550 ms, every 50 ms
times = np.arange(0.20, 0.55, 0.05)

plot_topomaps(cfg, window_name='trial',
              times=times.tolist(),
              conditions=['spatial', 'symbolic', 'spatial_vs_symbolic'],
              save=False)
```

```python
posterior_channels = ['Pz', 'P3', 'P4', 'P7', 'P8', 'Oz', 'O1', 'O2']

plot_channel_overlay_with_ci(
    cfg, cfg_erp,
    window_name='trial',
    channels=posterior_channels,
    conditions=['spatial', 'symbolic'],
    save=False,
)
```

```python
# Define ROIs inline — easy to tweak as you explore
rois = {
    'posterior_left':  ['P3', 'P7', 'O1'],
    'posterior_right': ['P4', 'P8', 'O2'],
    'posterior_mid':   ['Pz', 'Oz'],
}

plot_roi_overlay(cfg, window_name='cue', rois=rois,
                 conditions=['spatial', 'symbolic'],
                 with_ci=True, save=False)
```

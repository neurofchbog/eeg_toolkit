# ERP Factorial — 01: Compute Evokeds

Computes per-subject averages for all condition levels in a multi-factor design.
No trial equalization — too many conditions, too few trials per cell.

Conditions are defined in the factorial ERP config (e.g. `erp_position.yaml`).
Name them as `factor1_level_factor2_level` so they can be parsed into factors
in notebook 03 (e.g. `spatial_target_left`).

**Output:** `<subject>_<window>_<analysis_name>_ave.fif` per subject, group grand averages

```python
%load_ext autoreload
%autoreload 2

from eeg_toolkit import (
    load_config,
    find_subjects,
    compute_evokeds_subject,
    compute_evokeds_all,
    compute_grand_averages,
)

# ── Update these paths ──
cfg     = load_config('../../../configs/your_experiment.yaml')
cfg_erp = load_config('../../../configs/your_erp_factorial.yaml')

subjects = find_subjects(cfg)
print(f"Subjects:  {len(subjects)}")
print(f"Analysis:  {getattr(cfg_erp.erp, 'analysis_name', '(default)')}")
print(f"Windows:   {[w.name for w in cfg_erp.erp.windows]}")
print(f"Conditions ({len(vars(cfg_erp.conditions.cue))}):")  # update 'cue' to your window name
for name in vars(cfg_erp.conditions.cue):  # update 'cue'
    print(f"   {name}")
```

```python
# ── Test with first subject ──
test_subject = subjects[0]
ok, counts = compute_evokeds_subject(
    cfg, cfg_erp, test_subject,
    overwrite=True, verbose=True,
)
print(f"\nResult: {ok}")
```

```python
# ── Compute evokeds for all subjects ──
evoked_summary = compute_evokeds_all(cfg, cfg_erp, overwrite=False, verbose=True)
compute_grand_averages(cfg, cfg_erp, overwrite=True, verbose=True)
```

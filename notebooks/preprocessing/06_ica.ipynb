```python
%load_ext autoreload
%autoreload 2

import matplotlib
matplotlib.use('Qt5Agg')

from eeg_toolkit import (
    load_config,
    find_subjects,
    fit_ica_subject,
    label_ica_subject,
    inspect_ica_subject,
    apply_ica_subject,
)

cfg = load_config('../configs/eye_eeg_simul.yaml')
subjects = find_subjects(cfg)
print(f"Subjects: {len(subjects)}")
```

```python
test_subject = subjects[0]
print(f"\nFitting ICA for {test_subject}\n")

fit_ica_subject(cfg, test_subject, overwrite=False, verbose=True)
```

```python
print(f"Running ICLabel for {test_subject}\n")

label_ica_subject(cfg, test_subject, overwrite=True, verbose=True)
```

```python
# Open the interactive interface for the test subject
# This will open two windows: topographies and source time-courses
inspect_ica_subject(cfg, test_subject, verbose=True)
```

```python
from eeg_toolkit import fit_ica_all, label_ica_all

print("\n--- Step 1: Fitting ICA for all subjects ---")
# Fits the ICA model on the designated window (e.g., 'trial') for each subject.
# It uses a 1 Hz high-pass filtered copy of the data internally for stability.
fit_summary = fit_ica_all(cfg, overwrite=False, verbose=True)

print("\n--- Step 2: Running ICLabel for all subjects ---")
# Runs the neural network to classify components and pre-marks artifacts
# (e.g., eye blinks, muscle noise) based on the threshold in your YAML.
label_summary = label_ica_all(cfg, overwrite=False, verbose=True)
```

```python
from eeg_toolkit import inspect_ica_all

print("\n--- Step 3: Interactive ICA Inspection ---")
# Opens the topographies and source plots for each subject sequentially.
# - Red titles indicate components flagged by ICLabel.
# - Click topographies to toggle exclusion.
# - Close BOTH windows to save the final selection and automatically load the next subject.
inspect_summary = inspect_ica_all(cfg, verbose=True)
```

```python
from eeg_toolkit import apply_ica_all

print("\n--- Step 4: Applying cleaned ICA to all epochs ---")
# Projects out the manually confirmed excluded components from all epoching windows.
# Saves the results as <subject>_<window>_clean-epo.fif.
apply_summary = apply_ica_all(cfg, overwrite=False, verbose=True)
```

```python
# === Final verification ===
import json
from eeg_toolkit import load_status, find_subjects, get_clean_epochs_path
from eeg_toolkit.ica import _get_iclabel_path
import mne

cfg = load_config('../configs/eye_eeg_simul.yaml')
subjects = find_subjects(cfg)

print(f"=== ICA Pipeline Summary ({len(subjects)} subjects) ===\n")

all_ok = True
for subj in subjects:
    # Check clean epochs exist for all windows
    trial_path = get_clean_epochs_path(cfg, subj, 'trial')
    cue_path   = get_clean_epochs_path(cfg, subj, 'cue')
    iclabel_path = _get_iclabel_path(cfg, subj)
    
    trial_ok = trial_path.exists()
    cue_ok   = cue_path.exists()
    
    # Read ICLabel info
    n_excluded = "?"
    if iclabel_path.exists():
        with open(iclabel_path, 'r') as f:
            log = json.load(f)
        n_excluded = len([c for c in log['components'] if c['auto_excluded']])
    
    status = "OK" if (trial_ok and cue_ok) else "INCOMPLETE"
    if status != "OK":
        all_ok = False
    
    print(f"  [{status:10s}] {subj}: "
          f"ICLabel excluded={n_excluded}, "
          f"trial={'OK' if trial_ok else 'MISSING'}, "
          f"cue={'OK' if cue_ok else 'MISSING'}")

print(f"\n{'All subjects complete!' if all_ok else 'Some subjects incomplete — check above.'}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

group = load_all_scores(cfg, "cue", "spatial_vs_symbolic")
```

```python
# ── 1. Cue position: left vs right (SPATIAL — position is TARGET) ──
cond_pos_spatial = {
    "left":  [21, 31, 41, 12, 32, 42],
    "right": [13, 23, 43, 14, 24, 34],
}

# ── 2. Cue position: left vs right (SYMBOLIC — position is DISTRACTOR) ──
cond_pos_symbolic = {
    "left":  [121, 131, 141, 112, 132, 142],
    "right": [113, 123, 143, 114, 124, 134],
}

# ── 3. Cue digit: low vs high (SYMBOLIC — digit is TARGET) ──
cond_dig_symbolic = {
    "low":  [112, 113, 114, 121, 123, 124],
    "high": [131, 132, 134, 141, 142, 143],
}

# ── 4. Cue digit: low vs high (SPATIAL — digit is DISTRACTOR) ──
cond_dig_spatial = {
    "low":  [12, 13, 14, 21, 23, 24],
    "high": [31, 32, 34, 41, 42, 43],
}
```

```python
analyses = [
    ("position_spatial",  cond_pos_spatial,  "Position (target)"),
    ("position_symbolic", cond_pos_symbolic, "Position (distractor)"),
    ("digit_symbolic",    cond_dig_symbolic, "Digit (target)"),
    ("digit_spatial",     cond_dig_spatial,  "Digit (distractor)"),
]

common = dict(
    window_name="cue",
    classifier="linear_svc",
    n_folds=3,
    n_repeats=10,
    resample_sfreq=256,
    baseline=(-0.2, 0.0),
    n_null_permutations=1,
    n_jobs=-1,
)

for analysis_name, conditions, label in analyses:
    print(f"\n{'='*60}")
    print(f"  {label}")
    print(f"{'='*60}\n")
    summary, group = decode_all(
        cfg,
        conditions=conditions,
        analysis_name=analysis_name,
        **common,
    )
```

```python
from eeg_toolkit.mvpa import cross_decode_all

# Train on spatial (position = target), test on symbolic (position = distractor)
train_conds = {
    "left":  [21, 31, 41, 12, 32, 42],
    "right": [13, 23, 43, 14, 24, 34],
}
test_conds = {
    "left":  [121, 131, 141, 112, 132, 142],
    "right": [113, 123, 143, 114, 124, 134],
}

summary, group_xdec = cross_decode_all(
    cfg,
    window_name="cue",
    train_conditions=train_conds,
    test_conditions=test_conds,
    analysis_name="position_cross",
    classifier="linear_svc",
    resample_sfreq=256,
    baseline=(-0.2, 0.0),
    bidirectional=True,
    n_null_permutations=1,
    n_jobs=-1,
    overwrite = True
)
```

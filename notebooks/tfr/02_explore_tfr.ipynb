```python
# Loads the configuration and the function to visualize the data
%load_ext autoreload
%autoreload 2
from eeg_toolkit import (
    load_config,
    load_all_grand_averages,
    load_grand_average_tfr,
    plot_tfr_heatmap,
    plot_tfr_conditions,
    plot_tfr_topomaps,
    plot_tfr_timecourse,
)
cfg = load_config('../../configs/eye_eeg_simul.yaml')
cfg_tfr = load_config('../../configs/tfr_analysis.yaml')
print("Setup OK")
```

```python
# Here we are loading the grand average, its easier to visualize than having to load all subjects and then 
# calculating the average of all subjects.
ga_cue = load_all_grand_averages(cfg, cfg_tfr, "cue")
ga_trial = load_all_grand_averages(cfg, cfg_tfr, "trial")
print("Cue:", list(ga_cue.keys()))
print("Trial:", list(ga_trial.keys()))
```

```python
# This plot allows to look at all frequency activity in the conditions and clusters according to the electrodes chosen
plot_tfr_conditions(cfg, cfg_tfr, "trial", channels=["Pz", "Oz"],
                    conditions=["spatial", "symbolic", "spatial_vs_symbolic"]);
```

```python
# This plot allows us to look at the frequency activity 
plot_tfr_topomaps(cfg, cfg_tfr, "trial",
                  conditions=["spatial", "symbolic", "spatial_vs_symbolic"],
                  fmin=4, fmax=8, tmin=0.0, tmax=5);
```

```python
# Same as the code above
plot_tfr_topomaps(cfg, cfg_tfr, "cue",
                  conditions=["spatial", "symbolic", "spatial_vs_symbolic"],
                  fmin=8, fmax=13, tmin=0.3, tmax=0.6);
```

```python
# In this case we are plotting the average of the time frequency 
plot_tfr_timecourse(cfg, cfg_tfr, "trial", channels=["Pz", "Oz"],
                    fmin=8, fmax=12, conditions=["spatial", "symbolic"],
                    tmin = -0.1, tmax = 5);
```

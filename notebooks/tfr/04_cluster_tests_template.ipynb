# TFR — 04: Spatiotemporal Cluster Tests

Runs cluster-based permutation tests across channels × time for a specific
frequency band. This is the data-driven approach: no ROI or time window
pre-selection needed.

Run this after exploring in notebooks 02–03 to confirm effects spatiotemporally.

```python
%load_ext autoreload
%autoreload 2

from eeg_toolkit import (
    load_config,
    get_eeg_channels,
    build_adjacency,
    load_band_data,
    run_cluster_test,
    describe_clusters,
    plot_cluster_results,
)

# ── Update these paths ──
cfg     = load_config('../../configs/your_experiment.yaml')
cfg_tfr = load_config('../../configs/your_tfr_analysis.yaml')

print("Setup OK")
```

```python
# ── Get channel list and build adjacency ──
eeg_channels = get_eeg_channels(cfg, cfg_tfr, 'your_window')
print(f"EEG channels: {len(eeg_channels)}")
```

```python
# ── Load data for a specific frequency band ──
# ── Update freq_band, tmin, tmax to match your analysis ──
data, times, subjects = load_band_data(
    cfg, cfg_tfr,
    window_name='your_window',
    freq_band=(8, 13),     # alpha band — update as needed
    eeg_channels=eeg_channels,
    tmin=-0.2, tmax=1.0,
)
adjacency = build_adjacency(eeg_channels, len(times))
print(f"Data shape per condition: {data['condition_a'].shape}")
```

```python
# ── Run cluster-based permutation test ──
results = run_cluster_test(
    data['condition_a'], data['condition_b'],
    adjacency,
    label='Alpha: condition_a vs condition_b',
    n_permutations=1024,   # increase to 5000–10000 for publication
)
describe_clusters(results, eeg_channels, times)
```

```python
# ── Plot cluster results ──
plot_cluster_results(
    data['condition_a'], data['condition_b'],
    results, eeg_channels, times,
    label_a='Condition A',
    label_b='Condition B',
    tmin=-0.2, tmax=1.0,
)
```
